```python
source_root = "abfss://source-backlog@stadbenterprisede98.dfs.core.windows.net"
landing_root = "abfss://landing@stadbenterprisede98.dfs.core.windows.net"

yellow_files = sorted(f.name for f in dbutils.fs.ls(source_root + "/yellow"))
fhv_files = sorted(f.name for f in dbutils.fs.ls(source_root + "/fhv"))
fhvhv_files = sorted(f.name for f in dbutils.fs.ls(source_root + "/fhvhv"))
```

```python
yellow_info = dbutils.fs.ls(source_root + "/yellow")
fhv_info = dbutils.fs.ls(source_root + "/fhv")
fhvhv_info = dbutils.fs.ls(source_root + "/fhvhv")

total_bytes = (
    sum(file.size for file in yellow_info)
    + sum(file.size for file in fhv_info)
    + sum(file.size for file in fhvhv_info)
)

total_gb = total_bytes / (1024 ** 3)

print("NYC Taxi & Limousine Commission (TLC) Source Dataset Summary")
print("Yellow Taxi files:", len(yellow_info))
print("FHV files:", len(fhv_info))
print("HVFHV files:", len(fhvhv_info))
print("Total source dataset:", round(total_gb, 2), "GB")
```

```python
released_count = len(dbutils.fs.ls(landing_root + "/yellow"))

if released_count >= len(yellow_files):
    print("All monthly batches have already been released.")
    dbutils.notebook.exit("No new batch to release.")

next_file = yellow_files[released_count]

next_month = next_file.replace("yellow_tripdata_", "").replace(".parquet", "")

print("Next batch:", next_month)
```

```python
fhv_file = "fhv_tripdata_" + next_month + ".parquet"

if fhv_file in fhv_files:
    dbutils.fs.cp(
        source_root + "/fhv/" + fhv_file,
        landing_root + "/fhv/" + fhv_file
    )

    print("Copied:", fhv_file)
```

```python
fhvhv_file = "fhvhv_tripdata_" + next_month + ".parquet"

if fhvhv_file in fhvhv_files:
    dbutils.fs.cp(
        source_root + "/fhvhv/" + fhvhv_file,
        landing_root + "/fhvhv/" + fhvhv_file
    )

    print("Copied:", fhvhv_file)
else:
    print("No HVFHV file for:", next_month)
```

```python
dbutils.fs.cp(
    source_root + "/yellow/" + next_file,
    landing_root + "/yellow/" + next_file
)

print("Batch completed:", next_month)
```
